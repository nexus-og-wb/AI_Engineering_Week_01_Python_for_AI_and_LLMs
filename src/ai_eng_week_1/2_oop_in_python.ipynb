## 1. Classes and Object-Oriented Programming

Python supports Object-Oriented Programming (OOP) without requiring a `new` keyword for instantiation[cite: 1].

### Core Concepts
* `__init__`: Constructor method called when initializing an instance[cite: 1].
* `self`: Explicit reference to the current instance (passed automatically as the first parameter)[cite: 1].

> **Resource:** Tutorial on [Python Classes and Objects](https://www.w3schools.com/python/python_classes.asp)[cite: 1].

```python
class Dog:
    species = "Dog"

    def __init__(self, name: str, breed: str):
        self.name = name
        self.breed = breed

    def bark(self):
        self.eyes = 2
        return f"{self.name} says Woof and has {self.eyes} eyes!"

    @classmethod
    def walk(cls):
        print(f"{cls.species} is walking")
    
    @staticmethod
    def run():
        print("Dog is running")

    def thinking(self, thought: str):
        return f"{self.name} is thinking about {thought}"


Dog.walk()

Dog.run()

leo = Dog(name="Leo", breed="Labra")

print(leo.bark())

print(leo.eyes)

print(leo.breed)

print(leo.thinking("what to do next?"))
```

## 2. Inheritance and `super()`

Subclasses inherit attributes and methods from base classes[cite: 1]. Use `super()` to delegate initialization or invocation to parent classes[cite: 1].

> **Resource:** Read more about [Python Inheritance](https://www.w3schools.com/python/python_inheritance.asp).

```python
class Animal:
    def __init__(self, name: str):
        self.name = name

    def speak(self) -> str:
        return "..."

class Dog(Animal):
    def __init__(self, name: str, breed: str):
        super().__init__(name)           # Delegate initialization to Animal
        self.breed = breed

    def speak(self) -> str:              # Override method
        return "Woof!"

generic_animal = Animal("Some Animal")
rex = Dog("Rex", "Labrador")

print(f"generic_animal.speak() -> {generic_animal.speak()}")
print(f"rex.speak()            -> {rex.speak()}")
print(f"rex.name (from parent __init__) -> {rex.name}")
print(f"rex.breed (from Dog's own __init__) -> {rex.breed}")
```

## 2. Variadic Arguments: `*args` and `**kwargs`

Variadic parameters allow functions to accept an arbitrary number of arguments[cite: 1]. This design pattern is used across AI frameworks (e.g., PyTorch, LangChain, Hugging Face) for passing optional model configurations[cite: 1].

* `*args`: Collects positional arguments into a **tuple**[cite: 1].
* `**kwargs`: Collects keyword arguments into a **dictionary**[cite: 1].

[args and kwargs](https://www.geeksforgeeks.org/python/args-kwargs-python/)[cite: 1]

```python
def add_numbers(*args):
    print(type(args))
    print(args)

add_numbers(1, 2, 3, 4, 5, "Prashant", True, None)
```

```python
def user_profile(**kwargs):
    print(type(kwargs))
    print(f"name is {kwargs['name']}")
    print(kwargs)

user_profile(name="Prashant", age=22, email="prashant@example.com")
```

```python
def create_profile(name, role, experience):
    print(name)
    print(role)
    print(experience)

profile = {
    "name": "Prashant",
    "role": "SDE",
    "experience": 14
}

# profile = ["Prashant", "SDE", 14]

create_profile(name="Jaskarn", role="SDE", experience=5)
create_profile(**profile)
```

```python
def add(a, b, c):
    return sum((a, b, c))

k = [1, 2, 3]

print(add(*k))
```
